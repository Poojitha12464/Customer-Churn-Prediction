```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import os
import kagglehub

# Re-download path if necessary (ensures 'path' is defined in this cell's scope)
path = kagglehub.dataset_download("blastchar/telco-customer-churn")

# Re-load df if necessary (ensures 'df' is defined in this cell's scope)
df = pd.read_csv(os.path.join(path, "WA_Fn-UseC_-Telco-Customer-Churn.csv"))

df_encoded = pd.get_dummies(df, drop_first=True) # Re-defining df_encoded for robustness
plt.figure(figsize=(14,10))
sns.heatmap(df_encoded.corr(), cmap='coolwarm')
plt.show()
```
